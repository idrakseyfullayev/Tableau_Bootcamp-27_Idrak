## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders5_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_5_6.pdf'

# Başlıqlar (plt.title-lardan)
titles = [
    "Age-i Convert to Dimension etmək",
    "Dumbbell Chart",
    "Aggregate Measures 'Turn Off' etmək",
    "Year-i Color-ə atmaq, Ownership-i 2-ci Axis-ə də atmaq",
    "Axis-in birini Line Chart etmək və Dual Axis etmək",
    "Line Chart-da Year-i Color-dən çıxarıb Path-ə atmaq",
    "Ownership-i Line Chart-ın Label-na atmaq və Row-da Axis-lərin yerini dəyişmək",
    "Result",
    "Rəqəmlərin Normal görünməsi üçün - Create Calculated Field - 2105",
    "Rəqəmlərin Normal görünməsi üçün - Create Calculated Field - 2019",
    "2015 Calculated-i Circle-da Label-a atmaq",
    "2019 Calculated-i Line-da Label-a atmaq",
    "Axis üzərində Sağ Click - Format",
    "Axis - Scale - Numbers - Percentage",
    "2015 Calculated üzərində Sağ Click - Format",
    "Pane - Numbers - Percentage",
    "2019 Calculated üzərində Sağ Click - Format",
    "Pane - Numbers - Percentage",
    "Result"
]

# Arial fontunu qeyd et
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 78–96 şəkilləri PDF-ə əlavə edirik
for i in range(78, 97):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini saxla
    aspect = img.width / img.height
    available_height = height - 100
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 78
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli PDF-ə əlavə et
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifə
    c.showPage()

# PDF-i yadda saxla
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
