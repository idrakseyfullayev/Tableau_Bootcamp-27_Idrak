## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders5_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_5_Elave.pdf'

# Başlıqlar (plt.title-lardan)
titles = [
    "Data-nı JOIN etmək",
    "Number of Survey - Create Calculated Field",
    "Rating-i Convert To Dimension etmək",
    "Stacked Bar Chart",
    "Age - Create Calculated Field",
    "Histogram - Age Measure-dən Bins yaratmaq - Create - Bins...",
    "Histogram - Bins sayını seçmək",
    "Histogram - Age(bin)-i 'Convert to Continuous' etmək",
    "Histogram by Country and Age Analys",
    "ISNULL - Create Calculated Field",
    "Question-i Columns-a, Number of Survey-i Rows-a, Isnull Filter-ə atmaq",
    "Emojini Internet-dən yükləyib Emoji Qovluğu yaradıb qeyd olunan Path-ə yerləşdirmək",
    "Rating with Words - Create Calculated Field",
    "Happy or Sad - Create Calculated Field",
    "Emoji Chart",
    "Rating with Words - Sort - Nested - Ascending",
    "Nested - Ascending",
    "Dashboard - Action - Add Action - Filter",
    "Result"
]

# Arial fontunu qeyd et
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 97–115 şəkilləri PDF-ə əlavə edirik
for i in range(97, 116):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini saxla
    aspect = img.width / img.height
    available_height = height - 100
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 97
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli PDF-ə əlavə et
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifə
    c.showPage()

# PDF-i yadda saxla
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
