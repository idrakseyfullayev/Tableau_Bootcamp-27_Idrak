## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders5_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_5_3.pdf'

# Başlıqlar (plt.title()-lardan götürülüb)
titles_2 = [
    "İstənilən bir Measure-i Text-ə atmaq",
    "Text-də Measure-in əvəzinə adi text yazmaq",
    "Vizualı Floating kimi ekrana atırıq və təmizləyirik",
    "Dashboard - Action - Add Action - Go to Sheet",
    "'Source Sheets' - Hansı seçilsin, 'Target Sheets' - hansı Sheet-ə getsin",
    "Click üçün İkon yaratmaq - Measures sahəsində Double click",
    "Shape Chart - Shape - More Shapes - Edit Shape",
    "Size-ı artırmaq",
    "Vizualı Floating kimi ekrana atırıq və Dartıb böyüdürük",
    "Dashboard - Action - Hazır 'Go to Sheet' - Edit - Source Sheets - İkon Vizualı seçmək",
    "Dashboard - Action - Add Action - Go to URL",
    "'Source Sheets' - Pie seçmək, 'URL' - Link yazırıq - Hansı Genre növünə getsin",
    "'Menu' seçdiyimiz üçün Genre növünü seçəndə 'Menu'-da Link yaranır",
    "Link-ə click etməklə Web site-a keçid alırıq",
    "Dashboard - Show Grid",
    "Dashboard-ın Background-una Grid əlavə olunur"
]

# Arial fontu qeyd olunur
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 36–51 şəkilləri PDF-ə əlavə edirik
for i in range(36, 52):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 100  # bir az fərqli aralıq
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 36
    title = titles_2[title_index] if title_index < len(titles_2) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifə
    c.showPage()

# PDF-i yadda saxlayırıq
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
