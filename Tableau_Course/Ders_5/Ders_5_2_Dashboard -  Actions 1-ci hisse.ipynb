## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders5_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_5_2.pdf'

# Başlıqlar
titles = [
    "Dashboard",
    "Dashboard - Actions (Filter üçün)",
    "Add Action - Filter",
    "Lazımı qaydaları seçmək - Adi Filter qaydası",
    "Result",
    "Action - Filter - Exclude all values - üçün Line Chart",
    "Line Chart-ı Floating kimi ekrana atmaq",
    "Line Chart-ı təmizləmək",
    "Dashboard - Actions",
    "Add Action - Filter",
    "Clearing the selection will - Exclude all values",
    "Result - Seçimi ləğv etsək 'Line Chart Floating' də gizlənəcək",
    "Action-ı 'Edit' etmək və ya 'Delete' etmək",
    "Filteri-i seçib  'Edit' və ya 'Remove' etmək",
    "Dashboard - Action - Add Action - Highlihgt",
    "'Source Sheets' - Hansı Highlight olsun, 'Target Sheets' - hansı Solsun",
    "Result"
]

# Arial fontunu qeyd edirik
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 19–35 şəkilləri PDF-ə əlavə edirik
for i in range(19, 36):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 19
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifəyə keçirik
    c.showPage()

# PDF-i yadda saxlayırıq
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
