## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders5_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders5.pdf'

# Başlıqlar
titles = [
    "INDEX() - İndeks yaratmaq - Create Caluclated Field",
    "Calculated-i Discrete-ə Çevirmək",
    "Birinci mütləq Nəsə atılmalıdır Sonra Index",
    "CONTAINS() - Axtarış Funksiyası - Create Caluclated Field",
    "Contains Caluclated-i Filter-ə, Product Name-i Row-a",
    "CASE WHEN - Şərt Operatoru - Create Caluclated Field",
    "Case When Caluclated-i Dimension-a Çevirmək",
    "Case When-i Text-ə, Ship Mode-u Row-a",
    "2-ci - Üsul Case When-i Measure-ə qaytarmaq",
    "Case When-i Text-ə, Ship Mode-u Row-a",
    "Case When-i MAX və ya MİN etmək",
    "Result",
    "LOOKUP - Create Calculated Field",
    "1-ci Üsul - Year-i Row-a, Boys Participation və Lookup-ı 'Show Me' qaydası ilə",
    "2-ci Üsul - Year-i Row-a, Measure Names-i Columns-a, Measure Values-ı Text-ə",
    "Measure Names-i Filter-ə atmaq və Lazımı Sütünları seçmək",
    "Lookp-da 'Compute Using - Table' Seçmək",
    "Result"
]

# Arial fontunu qeyd edirik
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 1–18 şəkilləri PDF-ə əlavə edirik
for i in range(1, 19):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 1
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifəyə keçirik
    c.showPage()

# PDF-i yadda saxlayırıq
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
