## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders3_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders3_4_.pdf'

# Başlıqlar (plt.title hissələrindən götürülüb)
titles = [
    "İstifadəçi üçün 'Top 3' seçimi",
    "",
    "",
    "Seçilənlər içərisindən ancaq 'Top 3' saxlayır",
    "Create Parameter",
    "",
    "Show Parameter - İstifadəçi üçün (Amma işləmir çünki Parameter-i nəyəsə bağlamaq lazımdır)",
    "Parameter-ə bağlamaq üçün Set - Top_N yaratmaq",
    "",
    "",
    "Set - Top_N-u Color-ə atmaq (Filter-ə də atmaq olar)",
    "Parameter-ə bağlamaq üçün Filter - Top_N yaratmaq",
    "",
    "Result",
    "Başlıq Adını Parameter-ə bağlamaq",
    "Result"
]

# Arial fontunu qeyd edirik
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 44–59 şəkilləri PDF-ə əlavə edirik
for i in range(44, 60):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 44
    if title_index < len(titles):
        title = titles[title_index]
    else:
        title = ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    c.showPage()

# PDF-i tamamla
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
