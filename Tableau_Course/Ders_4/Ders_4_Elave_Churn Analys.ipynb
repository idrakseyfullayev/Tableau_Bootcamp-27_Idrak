## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders4_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Der_4_Elave.pdf'

# Başlıqlar
titles = [
    "Create Calculated Field - Total Churn sütunu yaratmaq",
    "Create Calculated Field - Churn Rate",
    "Churn Rate - Value Format - Faizə Çevirmək",
    "Value Format",
    "Edit Label - Value-nın qabağına Söz yazmaq",
    "Churn Rate May",
    "Churn Rate June",
    "Churn Rate Jule",
    "Edit Label with Tooltip Trips & Ticks - Copy edirik",
    "Edit Label - Tooltip-dən Copy edib bura Past edirik",
    "Measure Names & Measure Values (Bütün Measures adları və dəyərləri)",
    "Measure Names-i Color-ə atmaq",
    "Default Properties - Number Format - Percentage",
    "Percentage - İstifadə zamanı həmişə Percentage olacaq",
    "Total Churn Duplicated edib Adını dəyişdim",
    "Dimension-a çevirdim",
    "Total Churn Dimension to Row və Measure Names to Filter",
    "Measure Names & Measure Values məntiqi",
    "Box Plot",
    "Churn Analizi"
]

# Font qeydiyyatı
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradılması
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# Şəkilləri 106–125 aralığında əlavə edirik
for i in range(106, 126):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Başlıq
    title_index = i - 106
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkil nisbəti və ölçüsü
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Şəkilin yerləşməsi
    x = (width - new_width) / 2
    y = top_y - new_height

    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Səhifə nömrəsi
    c.setFont("Arial", 9)
    c.drawCentredString(width / 2, 20, f"Səhifə {i - 105}/{len(titles)}")

    c.showPage()

# PDF-i yadda saxla
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
