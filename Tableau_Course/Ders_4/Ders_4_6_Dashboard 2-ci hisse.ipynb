## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders4_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_4_6_.pdf'

# Başlıqlar
titles = [
    "Dashboard - Vizualları Dashboard-a atmaq",
    "Floating - Dashboard-a atılan hər şey Havada qalır",
    "Show Dashboard Title",
    "Dashboard Title Edit",
    "Chart Title-ı Hide etmək (Title üzərində sağ Click)",
    "Row Axis-i Hide etmək (Axis üzərində sağ Click)",
    "Row Lines-ı Hide etmək",
    "Həm Row Lines-ı, həm Column Lines-ı Hide etmək",
    "Row Divider və Column Divider-ı Hide etmək",
    "Zero Line-ı Hide etmək",
    "Layout - Border - Vizuallar və Title-a border əlavə etmək",
    "Sheet Title-a Border əlavə etmək",
    "Objects - Image - Dashboard-a Image əlavə etmək",
    "Image-i Floating-dən Konteyner-ə Çevirmək",
    "Objects - Web Page - Dashboard-a Sayt əlavə etmək",
    "Objects - Navigation - Bir Dashboard-dan digərinə keçid",
    "Bu halda işləmir - Ancaq Presentation Mode-a keçmək lazımdır",
    "Presentation Mode - Navigator-a Click et 'Dashboard 1'-ə keçid et",
    "Presentation Mode - Navigator-a Click et 'Dashboard 2'-ə keçid et"
]

# Arial fontunu qeyd edirik
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 73–91 şəkilləri PDF-ə əlavə edirik
for i in range(73, 92):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 73
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifəyə keçirik
    c.showPage()

# PDF-i yadda saxlayırıq
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
