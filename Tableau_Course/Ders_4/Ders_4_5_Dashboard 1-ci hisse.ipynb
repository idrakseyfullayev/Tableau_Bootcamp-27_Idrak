## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders4_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_4_5_.pdf'

# Başlıqlar
titles = [
    "Dashboard - Vizaulları Dashboard-a atmaq",
    "Dashboard Sahəsini Genişləndirmək - Size",
    "Dashboard-ı Clear etmək, Dublicate etmək, New WorkSheet yaratmaq",
    "Boş Konteyner yerləşdirmək (Sonra Vizual atmaq üçün)",
    "Objects - Blank - Vizuallar Arası boşluqlar yaratmaq üçün",
    "Objects - Text - Dashboard-a Text əlavə etmək",
    "1) Vizualı Remove etmək, 2) Vizualın olduğu Sheet-ə getmək",
    "Chart Title və Caption (Rəy yazılır) Passiv və ya Aktiv etmək",
    "Chart Title və Caption-nı Edit etmək (Double Click ilə)",
    "2-ci Dashboard üçün Vizuallar qurmaq - Calculated",
    "Pie Chart",
    "Quick Table Calculation - Percent of Total",
    "Value Format",
    "Value-ni Yuvarlaqlaşdırmaq",
    "Donut Chart",
    "Bar Chart",
    "Value Format",
    "Line Chart"
]

# Arial fontunu qeyd edirik
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 55–72 şəkilləri PDF-ə əlavə edirik
for i in range(55, 73):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 55
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifəyə keçirik
    c.showPage()

# PDF-i yadda saxlayırıq
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
