## PDF yaratmaq

```python
from PIL import Image
from reportlab.lib.pagesizes import A4
from reportlab.lib.utils import ImageReader
from reportlab.pdfgen import canvas
from reportlab.pdfbase.ttfonts import TTFont
from reportlab.pdfbase import pdfmetrics

# Fayl yolları
base_path = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Pictures\Ders4_'
output_pdf = r'C:\Users\dell\Desktop\Tableau\Tableau_Course\Ders_4_7_.pdf'

# Başlıqlar
titles = [
    "Chart-ın Filter-ni Aktiv etmək",
    "Digər Chart-ların 'Go to Sheet'-inə getsək Filter-də görə bilərik",
    "Result - Filter-də əlavə olunub",
    "Dashboard-un özünün Format-ı",
    "Dashboard-un Rəngini Dəyişmək",
    "Pick Screen Color ilə Dashboard-un Rəngini Saytdan Götürmək",
    "Sheet Title-ın Rəngini Dəyişmək",
    "Chart-ın Background Rəngini Dəyişmək - Pick Screen Color ilə",
    "Pick Screen Color",
    "Chart-ın Axis və Ticks Xətlərini Hide etmək",
    "Filter -  Harda istifadə olunmasını Seçmək (Sheet-də baş verir)",
    "Filter sonradan Sheet-də yaradılıbsa Dashboard-a əlavə etmək yolu",
    "Story - Dashboar-ları əlavə etmək, Size - Automatic, Presentation Mode",
    "Story - Presentation Mode - Slide"
]

# Arial fontunu qeyd edirik
pdfmetrics.registerFont(TTFont('Arial', r'C:\Windows\Fonts\arial.ttf'))

# PDF yaradırıq
c = canvas.Canvas(output_pdf, pagesize=A4)
width, height = A4

# 92–105 şəkilləri PDF-ə əlavə edirik
for i in range(92, 106):
    full_path = f'{base_path}{i}.jpg'
    img = Image.open(full_path)

    # Şəkil nisbətini hesablayırıq
    aspect = img.width / img.height
    available_height = height - 90
    available_width = width - 40
    new_width = available_width
    new_height = new_width / aspect

    if new_height > available_height:
        new_height = available_height
        new_width = new_height * aspect

    # Başlıq
    title_index = i - 92
    title = titles[title_index] if title_index < len(titles) else ""

    if title:
        c.setFont("Arial", 13)
        c.drawCentredString(width / 2, height - 40, title)
        top_y = height - 70
    else:
        top_y = height - 40

    # Şəkilin koordinatları
    x = (width - new_width) / 2
    y = top_y - new_height

    # Şəkli yerləşdir
    img_reader = ImageReader(img)
    c.drawImage(img_reader, x, y, width=new_width, height=new_height)

    # Yeni səhifə
    c.showPage()

# PDF-i yadda saxlayırıq
c.save()
print("✅ PDF yaradıldı:", output_pdf)
```
